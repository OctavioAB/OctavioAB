Give an algorithm to detect whether a given undirected graph contains a cycle. If the graph contains a cycle, then your algorithm should output one. (It should not output all cycles in the graph, just one of them.) The running time of your algorithm should be O(m + n) for a graph with n nodes and m edges.

```python
from dataclasses import dataclass
import warnings

e = [(2,1), (2,3), (4,2), (4,6), (6,5), (6,7)]
n = [1,2,3,4,5,6,7]

@dataclass
class GraphNode:
    name: int

    def __post_init__(self):
        self.id = self.name
    
    
class Graph:
    edges: list[tuple[GraphNode, GraphNode]]
    nodes: list[GraphNode]
    
    def __init__(self, edges: list, nodes: list):
        # We process the nodes first because that's why
        if not isinstance(nodes, list):
            raise TypeError("Argument 'nodes' must be a list.")
        
        self.nodes = []
        
        for data in nodes:
            if isinstance(data, GraphNode):
                self.nodes.append(data)
            elif not isinstance(data, int):
                raise TypeError(f"Argument 'nodes' must be a list of "
                                + f"GraphNodes or ints, found: {type(data)}")
            else:
                self.nodes.append(GraphNode(data))
        
        # Now we process the edges because of process of elimination why
        
        self.edges = []
        
        if not isinstance(edges, list):
            raise TypeError(f"Argument 'edges' must be a list.")
        
        for data in edges:
            from_node: GraphNode = None
            to_node: GraphNode = None
            if not isinstance(data, tuple | list):
                raise TypeError(f"Argument 'edges must be a list of tuples or lists")
            if (len(data)>0):
                if len(data) < 2:
                    raise RuntimeError(f"Argument 'edges' must be a list of tuple or lists "
                                       + f"of length at least 2: data {data}.")
                elif len(data)>2:
                    warnings.warn("Edge includes 'extra' data.")
                if not isinstance(data[0], GraphNode | int) or not isinstance(data[1], GraphNode | int):
                    raise TypeError(f"Argument 'edges' must be a list of tuples or lists "
                                    + f"of GraphNode or int. Found {type(data[0])} and {type(data[1])}")
                if isinstance(data[0], int):
                    from_node = GraphNode(data[0])
                else:
                    from_node = data[0]
                if isinstance(data[1], int):
                    to_node = GraphNode(data[1])
                else:
                    to_node = data[1]
                self.edges.append((from_node, to_node))
            else:
                warnings.warn(f"Empty 'edge' in initializer list.")

my_tree = Graph(edges= e, nodes= n)
print(my_tree.edges)
```
