# Greedy Algorithms
"An algorithm is greedy if it builds up a solution in small steps, choosing a decision at each step myopically to optimize some underlying criterion." <br>
"When a greedy algorithm succeeds in solving a nontrivial problem optimally, it typically implies something interesting and useful about the structure of the problem itself;" <br>
"It's easy to invent greedy algorithms for almost any problem; finding cases in which they work well, and proving that they work well, is the interesting challenge." <br>
"...two basic methods for proving that a greedy algorithm produces an optimal solution":<br>
1. the greedy algorithm stays ahead: in a step-by-step fashion, the greedy algorithm does better than any other algorithm at each step. 
2. echange argument: one considers any possible solution, and gradually transforms it into the greedy algorithm without hurting its quality.
<br>

"Most well-known applicationsof greedy algorithms: shortest paths in a graph, the minimum spanning tree problem, and the consturction of Huffman codes for performing data compression." <br>
"Finally, we consider a more complex application, the Minimum-Cost Arborescense Problem, which further extends our notion of what a greedy algorithm is."
