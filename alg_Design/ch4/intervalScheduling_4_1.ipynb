## 4.1 Interval Scheduling: The Greedy Algorithm Stays Ahead
"We have a set of requests {1, 2, …, n}; the i<sup>th</sup> request corresponds to an interval of time starting at s(i) and finishing at f(i)."
"We'll say that a subset of the requests is compatible if no two of them overlap in time, and our goal is to accept as large a compatible subset as possible. Compatible sets of maximum size will be called optimal."

### Designing a greedy algorithm
"The basic idea in a greedy algorithm for interval scheduling is to use a simple rule to select a first request i<sub>1</sub>. Once a request i<sub>1</sub> is accepted, we reject all requests that are not compatible with i<sub>2</sub>. We then select the next request i<sub>2</sub> to be accepted, and again reject all requests that are not compatible with i<sub>2</sub>. We continue in this fashion until we run out of requests." <br>
"The challenge in designing a good greedy algorithm is in deciding which simple ruse to use for the selection--and there are many natural rules for this problem that do not give good solutions." <br>

"A gready rule that does lead to the optimal solution (...): (...) accept first the request that finishes first, that is, the request _i_ for which _f(i)_ is as small as possible." <br>

"Let us state algorithm a bit more formally." <br>

```pseudocode
Initially let R be the set of all requests, and let A be empty

While R is not yet empty
    Choose a request i ∊ R that has the smallest finishing time
    Add request i to A
    Delete all requests from R that are not compatible with request i

EndWhile

Return the set A as the set of accepted requests
```

### Analyzing the algorithm
"As a start, we can immediately declare that the intervals in the set _A_ returned by the algorithm are all compatible." <br>

    (4.1) _A is a compatible se of requests_ <br>

"For the purposes of comparison, let _O_ be an optimal set of intervals. Ideally one might want to show that A = O, but this is too much to ask." <br>
"We will simply show that |_A_| = |_O_|, that is, that _A_ contains the same number of intervals as _O_ and hence is also an optimal solution." <br>

"The idea underlying the proof, as we suggested initially, will be to find a sense in which our greedy algorithm "stays ahead" of this solution _O_." <br>
"We will compare the partial solutions that the greedy algorithm constructs to initial segments of the solution _O_, and how that the greedy algorithm is doing better in a step-by-step fashion." <br>
